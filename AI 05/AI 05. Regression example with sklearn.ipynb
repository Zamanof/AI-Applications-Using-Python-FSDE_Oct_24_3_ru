```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
```

```python
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([53, 59, 65, 72, 76])
```

```python
print(X)
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
a = model.intercept_
b = model.coef_[0]
```

```python
print(f'Уравнение y = {a:.2f} + {b:.2f} * x')
```

```python
y_pred = model.predict(X)
y_pred
```

```python
plt.scatter(X, y, color='silver', label='Datas', zorder=2)
plt.plot(X, y, color='gold', label='Linear Regression', zorder=1)
plt.xlabel('Hours')
plt.ylabel('Marks')
plt.legend()

plt.show()
```
