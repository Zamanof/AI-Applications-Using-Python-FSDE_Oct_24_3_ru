```python
import matplotlib.pyplot as plt
```

```python
centuries = ["XI", "XII", "XIII", "XIV", "XV",
             "XVI", "XVII", "XVIII", "XIX", "XX"]

population_millions = [400, 440, 460, 440, 500,
                       520, 590, 990, 1630, 6150]
```

```python
plt.bar(centuries, population_millions, color="black")
plt.xlabel("Centuries")
plt.ylabel("Population")
plt.title("Population by Centuries")
plt.show()
```

```python
plt.bar(centuries, population_millions, color= ["red", "blue", "green", "yellow", "orange",
          "purple", "pink", "brown", "gray", "cyan"])
plt.xlabel("Centuries")
plt.ylabel("Population")
plt.title("Population by Centuries")
plt.show()
```

```python
plt.barh(centuries, population_millions, color= ["red", "blue", "green", "yellow", "orange",
          "purple", "pink", "brown", "gray", "cyan"])
plt.xlabel("Centuries")
plt.ylabel("Population")
plt.title("Population by Centuries")
for i, value in enumerate(population_millions):
    plt.text(i, i-0.1,str(value), ha='left', fontweight='bold', fontsize=8, color="black")
plt.show()
```
