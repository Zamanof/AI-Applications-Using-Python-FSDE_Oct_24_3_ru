```python
import matplotlib.pyplot as plt
```

```python
labels = ['Jan', 'Feb', 'Mar']
percents = [18, 25, 30]
```

```python
plt.pie(percents, labels=labels)
plt.title('Pie Diagram')
plt.show()
```

```python
plt.pie(percents, labels=labels, autopct='%1.2f%%')
plt.title('Pie Diagram')
plt.show()
```

```python
explode = [0, 0.1, 0]
plt.pie(
    percents,
    labels=labels,
    autopct='%1.2f%%',
    explode=explode,
    shadow=True,
    colors=['#24ced1', (0.94, 1, 0.17, 1), 'blue'],
    startangle=45
)
plt.title('Pie Diagram')
plt.show()
```
