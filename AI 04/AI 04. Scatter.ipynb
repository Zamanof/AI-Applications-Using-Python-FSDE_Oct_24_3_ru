### Визуальный просмотр всех карт

```python
import numpy as np
import matplotlib.pyplot as plt

# Основные карты без обратных версий
names = [name for name in plt.colormaps() if not name.endswith("_r")]

gradient = np.linspace(0, 1, 256).reshape(1, -1)

fig, axes = plt.subplots(
    nrows=len(names),
    figsize=(10, len(names) * 0.3)
)

for ax, name in zip(axes, names):
    ax.imshow(gradient, aspect="auto", cmap=name)
    ax.set_ylabel(
        name,
        rotation=0,
        ha="right",
        va="center"
    )
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()
plt.show()
```

### Пользовательские маркеры

| Формат | Описание | Пример |
|---|---|---|
| `"$...$"` | Символ или формула MathText | `marker=r"$\alpha$"` |
| `(numsides, 0, angle)` | Правильный многоугольник | `marker=(5, 0, 0)` |
| `(numsides, 1, angle)` | Звезда | `marker=(5, 1, 0)` |
| `(numsides, 2, angle)` | Лучи из центра | `marker=(5, 2, 0)` |
| Массив координат `(N, 2)` | Собственная форма | `marker=vertices` |
| Объект `Path` | Собственный контур | `marker=path` |
| Объект `MarkerStyle` | Маркер с дополнительными настройками | `marker=MarkerStyle("o")` |

- `numsides` — количество сторон или лучей.
- `angle` — угол поворота в градусах.

### Пример использования

```python
import matplotlib.pyplot as plt

plt.plot(
    [1, 2, 3, 4],
    [2, 5, 3, 7],
    marker="o",
    markersize=10,
    markerfacecolor="yellow",
    markeredgecolor="black",
    markeredgewidth=2
)

plt.show()
```

### Варианты заполнения для `plt.plot()`

Параметр `fillstyle`:

- `"full"` — полное заполнение.
- `"left"` — левая половина.
- `"right"` — правая половина.
- `"bottom"` — нижняя половина.
- `"top"` — верхняя половина.
- `"none"` — без заполнения.

```python
plt.plot(
    [1, 2, 3],
    [2, 4, 3],
    marker="o",
    markersize=15,
    fillstyle="left",
    markerfacecolor="blue",
    markerfacecoloralt="yellow"
)

plt.show()
```
