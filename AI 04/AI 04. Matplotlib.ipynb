```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(0, 10)
x
```

```python
y = x*2
y
```

```python
plt.plot(x, y)
plt.title('y = x^2')
plt.xlabel('x')
plt.ylabel('y')
plt.xlim(0, 10)
plt.ylim(0, 20)
plt.savefig('test.png', dpi=300)
plt.show()
```

```python
fig = plt.figure()
axes = fig.add_axes((0, 0, 1, 1)) # left, bottom, width, height
axes.plot(x, y)
plt.show()
```

```python
a = np.linspace(0, 10, 15)
a
```

```python
b = a ** 2
b
```

```python
fig = plt.figure()
axes1 = fig.add_axes((0, 0, 1, 1))
axes1.plot(a, b)
axes1.set_title('b = a ^ 2')
axes1.set_xlabel('a')
axes1.set_ylabel('b')
axes1.set_xlim(0, 10)
axes1.set_ylim(0, 100)

axes2 = fig.add_axes((0.5, 0.5, 0.5, 0.5)) # left, bottom, width, height
axes2.plot(x, y)
axes2.set_title('y = 2x')
axes2.set_xlabel('x')
axes2.set_ylabel('y')

plt.show()
```

```python
fig, axes = plt.subplots(nrows=1, ncols=2)
axes[0].plot(a, b)
axes[1].plot(x, y)
plt.tight_layout()

plt.show()
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2)
axes[0, 0].plot(a, b)
axes[1, 1].plot(x,y)
axes[1, 0].plot(x, np.sin(x))
axes[0, 1].plot(x, np.tan(x))
plt.tight_layout()
plt.show()
```

```python
fig = plt.figure()
axes = fig.add_axes((0, 0, 1, 1))
axes.plot(x, y, label='y=2x', color='#24ced1')
axes.plot(a, b, label='b=a^2', color='#be46e3', lw=3)
axes.plot(a, 10*np.sin(a), label='a=sin(a)*10', color='#f1ff2b', lw=5)
axes.legend()
plt.show()
```
