# Граф вычислений
$$
y = x^2 + 5x + 7
$$



1. Посчитать $x^2$
2. Посчитать $5x$
3. Сложить этот результат и добавить 7

```python
X = torch.tensor(2., requires_grad=True)
y = X**2 + 5*X + 7
X, y
```

```python
y.grad_fn
```

$$
y = x^2 + 5x + 7
$$

$$
\frac{dy}{dx} = 2x + 5
$$

$$
\frac{dy}{dx} = 2 ⋅ 2 + 5 = 9
$$

```python
y.backward()
print(X.grad)
```
